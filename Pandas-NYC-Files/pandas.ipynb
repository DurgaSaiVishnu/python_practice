```python
data["age"].fillna(data["age"].mean())
```

```python
data.drop_duplicates(subset="employee_id",keep="first")
data.duplicated().sum()
```

```python
data["joining_date"] = pd.to_datetime(data["joining_date"])
data.info()
```

```python
data.rename(columns={"name":"Name"})
```

```python
data["salary"] = pd.to_numeric(data["salary"],errors="coerce")
data["annual salary"] = data["salary"]*12
```

```python
def categorize(num):
    if num > 5:
        return "expert"
    else:
        return "learner"
```

```python
data["status"] = data["experience"].apply(categorize)
data["perfomance_category"] = data["performance_score"].apply(lambda x: "excellent" if x>6 else("good" if x>4 else "needs improvement"))
data
```

```python
raw_codes = {"engineering":"ENG","datascience":"DS","HR":"Hr","finance":"FN","marketing":"MK","Sales":"SL"}

codes = {k.lower().replace(" ",""):v for k, v in raw_codes.items()}

data["Department_codes"] = data["department"].str.replace(" ","").str.lower().map(codes)
data
```

```python
data.reset_index()
data.sort_values(by="salary",ascending=False)
```

```python
data["experience_rank"] = data["experience"].rank(ascending=False,method="min")
data.sort_values("experience_rank")
```
